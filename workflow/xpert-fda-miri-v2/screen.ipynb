# MIRI Drug Repositioning Screen — xpert-fda-miri-v2

Union-CM finetune sibling of the finalized xpert-fda-miri-v1 screen. Only FDA drugs present in the L1000 embedding table (exact SMILES match, drug_idx < 8981) are screened; the HG biological embeddings are the **author reference table** — no HG training / FDA-extension inference. The XPert finetune (steps 05/06) trains on the workflow-built **union h5ad of the primary + iPSC datasets** (step 03, ADR 0015) with real-concentration dose bins, a single-cell encoding, and per-plate normalization (ADR 0016); the benchmark test sets are byte-identical to v1.1's. Run cells in order. Environment: `conda activate arc`. Heavy steps (05 benchmark, 06 finetune, 07 predict) require a GPU.

Outputs: `01_drug_embed/output/drug_list.csv` (1,004 drugs) + `all_drugs_idx2HG.npy`.

## Step 03 — CM finetune-data union (the v2 change)

Reads the two finetune datasets read-only (`dtoxs_primary` from the LINCS LDS-1593 well-level package, `dtoxs_ipsc` from GSE217423) and assembles one per-dataset h5ad + the union `03_cm_finetune_data/output/cm_union.h5ad` (958 rows = 662 + 296, HG-resolved compounds only, 978 landmarks, matched per-sample controls, author dose bins from **real uM concentrations** — paper Table 1 KIs / iPSC series-matrix shared non-KIs / literature Cmax primary-only drugs, ADR 0015). Per-`(dataset, plate)` QN + matched controls since ADR 0016. Also emits the drug manifest, per-dataset gene maps, the delta report vs v1.1 `finetune_delta.md`, and the composition report `finetune_report.html`. The schema gate asserts the union is structure-identical to v1's `workflow/xpert-fda-miri-v1/03_cm_finetune_data/output/dtoxs.h5ad` in the fields the XPert loader consumes.

## Step 05 — Benchmark XPert on the union CM splits

The L1000 pretrain weight is zero-shot-evaluated as the first (amber) column. Per-fold values are overlaid as black dots on the mean ± std bars. Outputs: `05_xpert_finetune_bench/output/benchmark.csv` + `benchmark.png`.
