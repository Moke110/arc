## Step 05 — Benchmark XPert on the union CM splits (full/ + data-ablation)

Two independent benchmark methods (ADR 0018/0019). **full/** is the official fold benchmark: imports v1.1's fold columns verbatim for the primary arm (byte-identical test sets, ADR 0015) and assigns iPSC rows under the isolation rules (warm-start all-train; cold-drug test drugs excluded from both datasets' train) + `cm_full` (all-train), then finetunes each fold from the pretrain weight and evaluates on the held-out test set (4 epochs). The L1000 pretrain weight is zero-shot-evaluated as the first (amber) column. Outputs: `05_xpert_finetune_bench/full/output/benchmark.csv` + `full/benchmark.png`.

**data-ablation/** is the finetune-corpus scale diagnostic (ADR 0019, GPU, 30 finetune + 3 pretrain runs): drug-partitioned tripartition of the 45 union drugs, ~15% unified test per seed, 5 train-size fractions {20…100%} in warm-start and cold-drug modes → 4×4 `data-ablation/ablation.png` + `ablation.csv` + `ablation_report.md`.

## Step 12 — Top-20 evidence survey (agentic, external biomni)

`12_drug_survey/` surveys the step-11 **top-20 ranked drugs**: one biomni agent per drug (bilingual brief + two-score rubric in `12_drug_survey/prompt.md`) researches 中文名, NMPA + FDA approval, pharmacological classification, MI/MIRI evidence and other ischemia-reperfusion-disease evidence (sources cited inline; agent proposes `score_miri` + `score_iri`, step owner confirms). Outputs: `12_drug_survey/drugs/{drug}.md` (bilingual dossier per drug) + `12_drug_survey/reports@20.html` (summary table + embedded dossiers, built by `12_drug_survey/assemble_reports.py`). Step 12 is read-only on steps 01–11; the agent run needs the biomni backend and is NOT a screening step — run the agents outside this notebook, then rebuild the HTML with `!python 12_drug_survey/assemble_reports.py`.

```python
!python 12_drug_survey/assemble_reports.py
import pandas as pd
from pathlib import Path
ranked = pd.read_csv('11_integrate/output/ranked_drugs.csv')
top = ranked.sort_values('integrated', ascending=False).head(20)
drugs = [Path('12_drug_survey/drugs') / (d + '.md')
         for d in top['drug_name']]
done = [p for p in drugs if p.exists()]
print(f'top-20 survey dossiers: {len(done)}/{len(drugs)} present')
print('final report: 12_drug_survey/reports@20.html')
```
